# Bayesian Playground
This package contains modules and functions related to Bayesian analysis in the context of the Fermi Paradox.


```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
# Example function to demonstrate Bayesian updating
def bayesian_update(prior, likelihood, evidence):
    """
    Perform Bayesian update.
    
    Parameters:
    prior (float): Prior probability P(H)
    likelihood (float): Likelihood P(E|H)
    evidence (float): Evidence P(E)
    
    Returns:
    float: Posterior probability P(H|E)
    """
    posterior = (likelihood * prior) / evidence
    return posterior    

# Example usage
if __name__ == "__main__":
    prior = 0.5  # Initial belief
    likelihood = 0.8  # Probability of evidence given hypothesis
    evidence = 0.6  # Total probability of evidence
    
    posterior = bayesian_update(prior, likelihood, evidence)
    print(f"Posterior Probability: {posterior}")
```

```python
# Example sample plot

samples = np.random.beta(8, 2, size=100_000)

# Plotting the samples
plt.hist(samples, bins=50, density=True, alpha=0.7, color='blue')
plt.title('Belief about f_p (α=8, β=2)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.grid(True)
plt.show()
```

```python
# mean and std of samples
mean = np.mean(samples)
std = np.std(samples)
print(f"Mean: {mean}, Std Dev: {std}")
```
